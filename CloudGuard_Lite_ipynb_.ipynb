```python


import json
from datetime import datetime, timezone

def check_cloud_security(config):
    checks = [
        ("public_storage", "Storage must not be public", 25),
        ("encryption", "Storage encryption must be enabled", 25),
        ("mfa", "Multi-factor authentication must be enabled", 20),
        ("least_privilege", "Least-privilege access must be used", 20),
        ("logging", "Security logging must be enabled", 10),
    ]

    findings = []
    score = 100

    for key, message, penalty in checks:
        if config.get(key) is not (False if key == "public_storage" else True):
            score -= penalty
            findings.append({
                "check": key,
                "status": "FAIL",
                "message": message,
                "points_lost": penalty
            })
        else:
            findings.append({
                "check": key,
                "status": "PASS",
                "message": "Configuration check passed",
                "points_lost": 0
            })

    return {
        "project": "CloudGuard Lite",
        "checked_at": datetime.now(timezone.utc).isoformat(),
        "security_score": max(0, score),
        "risk_level": (
            "HIGH" if score < 50 else
            "MEDIUM" if score < 80 else "LOW"
        ),
        "findings": findings,
        "note": "Demo assessment, not a real cloud audit."
    }

# Example configuration with security weaknesses
sample_config = {
    "public_storage": True,
    "encryption": True,
    "mfa": False,
    "least_privilege": True,
    "logging": False
}

report = check_cloud_security(sample_config)
print(json.dumps(report, indent=2))
```

```python


# Automated tests for CloudGuard Lite

secure_config = {
    "public_storage": False,
    "encryption": True,
    "mfa": True,
    "least_privilege": True,
    "logging": True
}

insecure_config = {
    "public_storage": True,
    "encryption": False,
    "mfa": False,
    "least_privilege": False,
    "logging": False
}

secure_report = check_cloud_security(secure_config)
insecure_report = check_cloud_security(insecure_config)

assert secure_report["security_score"] == 100
assert secure_report["risk_level"] == "LOW"
assert insecure_report["security_score"] == 0
assert insecure_report["risk_level"] == "HIGH"
assert len(secure_report["findings"]) == 5
assert len(insecure_report["findings"]) == 5

print("PASS: Secure configuration test")
print("PASS: Insecure configuration test")
print("PASS: Five security checks verified")
print("All automated tests passed.")
```
